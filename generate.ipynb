```python
from templater import Item
from templater import Email
from templater import generate
```

```python
# Write all the stuff here, rembember that the below text is HTML, so you can
# use HTML tags if you want to.
title = f"The WAI Print"
subtitle = f"Week 3"
first_line = f"Week 3 already? Time flies when you're having fun, rightttt?"
footer = f"sent by a much sleep deprived WAI President &lt;3"


announcement_title = f"Announcements"
announcements = [
    Item(
        title=f"Fresher Reps",
        subtitle=f"Coming Soon!",
        content=(
            f"""
            We're looking for freshers to represent the WAI community.
            <br />
            <br />
            If you're interested, keep an eye out for the application form in
            the coming weeks.
            """
        )
    ),
    Item(
        title=f"Projects Launching",
        subtitle=f"Soon!",
        content=(
            f"""
            Alex will be launching the first project of the year soon, with more
            being released over the term!.
            """
        )
    ),
]

events_title = "Events"
events = [
    Item(
        title=f"WAI Cafe",
        subtitle=f"Tue 6-8pm @ FAB2.32",
        content=(
            f"""
            Join us for a social in the WAI Discord server! We'll be playing
            games and chatting, so come along and say hi!
            """
        )
    ),
    Item(
        title=f"CodeNight",
        subtitle=f"Thur 6-8pm @ FAB2.32",
        content=(
            f"""
            We're excited to announce that we'll be hosting a workshop with
            Microsoft on Thursday 7th October at 6pm.
            """
        )
    ),
]
```

```python
result = generate(
    Email(
        title=title,
        subtitle=subtitle,
        first_line=first_line,
        announcement_title=announcement_title,
        announcements=announcements,
        events_title=events_title,
        events=events,
        footer=footer
    )
)

with open("output.html", "w") as f:
  f.write(result)

# Open the file in your browser to see the result
import webbrowser

webbrowser.open("output.html")
```
